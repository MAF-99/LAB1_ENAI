```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.optim import Adam
from torch.utils.data import DataLoader
from torchvision.datasets import FashionMNIST
from torchvision.transforms import ToTensor

from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
import numpy as np
```

```python
from torchvision import datasets, transforms

# Define the transform to convert PIL images to PyTorch Tensors
transform = transforms.ToTensor()

# Class names for Fashion-MNIST (we'll use them for plotting)
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

# TODO: Create the training dataset
train_dataset = datasets.FashionMNIST(
    root='./data',        # Dataset storage path (e.g. './data')
    train=True,           # Use training data
    transform=transform,  # Convert images to tensors
    download=True         # Download if not available
)

# TODO: Create the testing dataset
test_dataset = datasets.FashionMNIST(
    root='./data',
    train=False,
    transform=transform,
    download=True
)

# Print one sample from the dataset
sample_image, sample_label = train_dataset[0]

print(f"Image shape: {sample_image.shape}")   # Expected: (1, 28, 28)
print(f"Label: {sample_label} -> {class_names[sample_label]}")
```

```python
from torch.utils.data import DataLoader

# TODO: DataLoader for training data
train_loader = DataLoader(
    train_dataset,
    batch_size=32,
    shuffle=True
)

# TODO: DataLoader for test/validation data
test_loader = DataLoader(
    test_dataset,
    batch_size=32,
    shuffle=False
)

# Get the first batch from the training DataLoader
X_batch, y_batch = next(iter(train_loader))
print(f"Training batch input shape: {X_batch.shape}")   # Expected: (32, 1, 28, 28)
print(f"Training batch labels shape: {y_batch.shape}")  # Expected: (32,)
```

```python

# Get one batch of images and labels
images, labels = next(iter(train_loader))

plt.figure(figsize=(8, 4))

for i in range(6):
    plt.subplot(2, 3, i + 1)
    # TODO: show the image in grayscale (hint: .squeeze() + cmap='gray')
    plt.imshow(images[i].squeeze(), cmap='gray')
    # TODO: set the title to the class name of the label
    plt.title(f"Label: {class_names[labels[i]]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python

class NN3Layer(nn.Module):

    def __init__(self, input_dim, hidden_dim, output_dim):
        super(NN3Layer, self).__init__()

        # TODO: First linear layer: input features -> hidden layer
        self.layer1 = nn.Linear(input_dim, hidden_dim)

        # TODO: Second linear layer: hidden layer -> hidden layer
        self.layer2 = nn.Linear(hidden_dim, hidden_dim)

        # TODO: Output layer: hidden layer -> number of classes (logits)
        self.layer3 = nn.Linear(hidden_dim, output_dim)

        # TODO: ReLU activation for non-linearity
        self.relu = nn.ReLU()

    def forward(self, x):
        # TODO: First hidden layer (linear -> relu)
        z1 = self.layer1(x)
        a1 = self.relu(z1)

        # TODO: Second hidden layer (linear -> relu)
        z2 = self.layer2(a1)
        a2 = self.relu(z2)

        # TODO: Output layer (raw scores / logits)
        output = self.layer3(a2)

        return output
```

```python
def train_one_epoch(model, optimizer, criterion, train_loader, device):
    # Set the model to training mode
    model.train()

    running_loss = 0.0

    for X_batch, y_batch in train_loader:
        # TODO: Flatten the batch and move it to the device
        X_batch = X_batch.view(X_batch.size(0), -1).to(device)
        y_batch = y_batch.to(device)

        # TODO: Forward pass
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)

        # TODO: Backward pass & optimization (3 steps)
        optimizer.zero_grad()   # Clear previous gradients
        loss.backward()         # Compute gradients
        optimizer.step()        # Update model parameters

        running_loss += loss.item()

    # Average loss over all batches
    avg_loss = running_loss / len(train_loader)

    return avg_loss
```

```python
def validate(model, criterion, test_loader, device):
    # TODO: Set the model to evaluation mode
    model.eval()

    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            # TODO: Flatten the batch and move it to the device
            X_batch = X_batch.view(X_batch.size(0), -1).to(device)
            y_batch = y_batch.to(device)

            # TODO: Forward pass
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            running_loss += loss.item()

            # TODO: Apply Softmax to get probabilities
            probabilities = torch.softmax(outputs, dim=1)

            # TODO: Multiclass predictions (argmax)
            predicted = torch.argmax(probabilities, dim=1)

            # Accuracy calculation
            correct += (predicted == y_batch).sum().item()
            total += y_batch.size(0)

    avg_loss = running_loss / len(test_loader)
    accuracy = correct / total

    return avg_loss, accuracy
```

```python
# Device setup
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# TODO: Model parameters
input_dim = 28 * 28   # Fashion-MNIST images flattened (784 features)
hidden_dim = 128      # Design choice (e.g., 128, 256, or 512 hidden units)
output_dim = 10       # 10 clothing classes

# TODO: Instantiate the model and move it to the device
model = NN3Layer(input_dim, hidden_dim, output_dim).to(device)

print("Model Architecture:\n")
print(model)

# Calculate the total number of trainable parameters
total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nTotal trainable parameters: {total_params}")
```

```python
num_epochs = 10
learning_rate = 0.001

# TODO: Define criterion (loss function)
criterion = nn.CrossEntropyLoss()

# TODO: Define optimizer (Adam)
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)
```

```python
# Run Training
train_losses = []
val_losses = []
val_accuracies = []

print('Starting Training...')
for epoch in range(num_epochs):
    # TODO: Train one epoch
    train_loss = train_one_epoch(model, optimizer, criterion, train_loader, device)

    # TODO: Validate
    val_loss, val_accuracy = validate(model, criterion, test_loader, device)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    val_accuracies.append(val_accuracy)

    print(f'Epoch [{epoch+1}/{num_epochs}], Train Loss: {train_loss:.4f}, Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}')

print('Training Complete!')
```

```python
# TODO: Fit PCA with n_components = 28*28
pca_full = PCA(n_components=28*28)
pca_full.fit(X)

# TODO: Calculate cumulative variance using np.cumsum on explained_variance_ratio_
cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(range(1, 785), cumulative_variance, 'bo-')
plt.axhline(y=0.80, color='r', linestyle='--', label='80% variance')
plt.axhline(y=0.95, color='g', linestyle='--', label='95% variance')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('PCA: Explained Variance vs Number of Components')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# TODO: Create 5 targets from 0.80 to 0.95 using np.linspace
targets = np.linspace(0.80, 0.95, 5)

for target in targets:
    # TODO: Find the first index where cumulative variance >= target (+1)
    n_components = np.argmax(cumulative_variance >= target) + 1

    print(f"To keep {target:.2%} variance, you need {n_components} components.")
```

```python
class Autoencoder(nn.Module):

    def __init__(self, latent_dim=32):
        super(Autoencoder, self).__init__()

        # TODO: Encoder: compress 784 -> latent_dim
        # 784 -> 256 -> 64 -> latent_dim (ReLU after each Linear)
        self.encoder = nn.Sequential(
            nn.Linear(784, 256),
            nn.ReLU(),
            nn.Linear(256, 64),
            nn.ReLU(),
            nn.Linear(64, latent_dim),
            nn.ReLU()
        )

        # TODO: Decoder: expand latent_dim -> 784
        # latent_dim -> 64 -> 256 -> 784 (ReLU between, Sigmoid at the end)
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 256),
            nn.ReLU(),
            nn.Linear(256, 784),
            nn.Sigmoid()
        )

    def encode(self, x):
        # x shape is (batch, 1, 28, 28)
        batch_size = x.size(0)

        # TODO: flatten to (batch, 784) then encode
        x = x.view(batch_size, -1)
        z = self.encoder(x)

        return z

    def decode(self, z):
        # z shape is (batch, latent_dim)
        batch_size = z.size(0)

        # TODO: decode then reshape back to (batch, 1, 28, 28)
        x_hat = self.decoder(z)
        x_hat = x_hat.view(batch_size, 1, 28, 28)

        return x_hat

    def forward(self, x):
        # TODO: encode then decode
        z = self.encode(x)
        x_hat = self.decode(z)
        return x_hat
```

```python
# Create model
LATENT_DIM = 32

# TODO: instantiate the Autoencoder
ae_model = Autoencoder(latent_dim=LATENT_DIM)

print("Autoencoder Architecture:")
print(ae_model)
```

```python
BATCH_SIZE = 128

# TODO: create train_loader (shuffle=True) and test_loader (shuffle=False)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

ae_model = ae_model.to(device)

# TODO: MSE Loss for reconstruction
criterion = nn.MSELoss()

# TODO: Adam optimizer with LEARNING_RATE = 0.001
LEARNING_RATE = 0.001
optimizer = torch.optim.Adam(ae_model.parameters(), lr=LEARNING_RATE)

NUM_EPOCHS = 10
```

```python
train_losses = []

for epoch in range(NUM_EPOCHS):
    ae_model.train()
    epoch_loss = 0.0

    for X_batch, _ in train_loader:  # We don't need y labels!
        X_batch = X_batch.to(device)

        # TODO: Forward pass
        X_hat = ae_model(X_batch)

        # TODO: Compute loss (compare original vs reconstruction)
        loss = criterion(X_hat, X_batch)

        # TODO: Backward pass (3 steps)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_loss)

    print(f"Epoch [{epoch+1:2d}/{NUM_EPOCHS}] | Loss: {avg_loss:.6f}")
```

```python
# Plot training loss
plt.figure(figsize=(10, 5))
plt.plot(train_losses, 'b-o')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.title('Autoencoder Training Loss')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
ae_model.eval()

test_images, test_labels = next(iter(test_loader))
test_images = test_images.to(device)

with torch.no_grad():
    # TODO: reconstruct the test images
    reconstructed = ae_model(test_images)

# Move to CPU for plotting
test_images = test_images.cpu()
reconstructed = reconstructed.cpu()

# Plot original vs reconstructed
n_images = 9
fig, axes = plt.subplots(2, n_images, figsize=(18, 6))

for i in range(n_images):
    # Original
    axes[0, i].imshow(test_images[i].squeeze(), cmap='gray')
    axes[0, i].axis('off')
    axes[0, i].set_title(f'Original\n"{class_names[test_labels[i]]}"', fontsize=9)

    # Reconstructed
    axes[1, i].imshow(reconstructed[i].squeeze(), cmap='gray')
    axes[1, i].axis('off')
    axes[1, i].set_title(f'Reconstructed\n"{class_names[test_labels[i]]}"', fontsize=9)

plt.suptitle('Original vs Reconstructed Images', fontsize=14)
plt.tight_layout()
plt.show()
```
